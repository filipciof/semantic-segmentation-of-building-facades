```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
from collections import Counter

from src.dataloader import get_train_test_val, get_labels, get_old_labels
```

```python
data = get_train_test_val()
X_train = data["train"]["images"]
y_train = data["train"]["masks"]
y_train_og = data["train"]["og_masks"]
```

```python
plt.imshow(X_train[0], vmin=1, vmax=12)
plt.show()
```

```python
im = plt.imshow(
    y_train_og[0], cmap="tab20", vmin=1, vmax=12
)

for i, name in enumerate(get_old_labels(), start=1):
    plt.plot([], [], "s", color=im.cmap(im.norm(i)), label=name)

plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```

```python
im = plt.imshow(
    y_train[0], cmap="tab20", vmin=1, vmax=12
)

for i, name in enumerate(get_labels(), start=1):
    plt.plot([], [], "s", color=im.cmap(im.norm(i)), label=name)

plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```
